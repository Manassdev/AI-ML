```python
from sympy import*
v1=[1,2,3]
v2=[4,5,6]
v3=[7,8,9]
M=Matrix([v1,v2,v3])
V=M.T
display(M)
Rank=V.rank()

if Rank==V.shape[1]:
    print("set is linearly independent")
else:
    print("set is linearly dependent")
Rank
```

```python
import numpy as np
v= np.array([[1,4,7],[2,5,8],[3,6,9]])
rank=np.linalg.matrix_rank(v)
print(rank)
if rank==V.shape[1]:
    print("set is linearly independent")
else:
    print("set is linearly dependent")
```

```python
from sympy import Matrix
M1 =Matrix([0,0,0,0])
M2 =Matrix([0,1,0,0])
M3 = Matrix([0,0,1,0])
A=Matrix([M1,M2,M3]).T

rank=A.rank()
print(rank)
if rank==A.shape[1]:
    print("set is linearly independent")
else:
    print("set is linearly dependent")
```
