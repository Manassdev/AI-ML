```python
import os
import base64
import requests
from getpass import getpass
from urllib.parse import quote

# GitHub settings
OWNER = "Manassdev"
REPO = "AI-ML"
FOLDER = "/content/drive/MyDrive/Colab Notebooks"

# Enter token securely
TOKEN = getpass("Enter your GitHub token: ")

headers = {
    "Authorization": f"Bearer {TOKEN}",
    "Accept": "application/vnd.github+json",
    "X-GitHub-Api-Version": "2022-11-28"
}

repo_url = f"https://api.github.com/repos/{OWNER}/{REPO}"

# Check repository and get its default branch
response = requests.get(repo_url, headers=headers)
response.raise_for_status()
branch = response.json()["default_branch"]

print("Repository:", REPO)
print("Branch:", branch)

# Upload all .ipynb files in the folder
files = [
    f for f in os.listdir(FOLDER)
    if f.lower().endswith(".ipynb")
    and os.path.isfile(os.path.join(FOLDER, f))
]

print("Notebooks found:", len(files))

for filename in files:
    filepath = os.path.join(FOLDER, filename)
    path = quote(filename, safe="")
    file_url = f"{repo_url}/contents/{path}"

    with open(filepath, "rb") as f:
        content = base64.b64encode(f.read()).decode("utf-8")

    # Check whether this file already exists
    existing = requests.get(
        file_url,
        headers=headers,
        params={"ref": branch}
    )

    payload = {
        "message": f"Upload notebook {filename}",
        "content": content,
        "branch": branch
    }

    if existing.status_code == 200:
        payload["sha"] = existing.json()["sha"]
    elif existing.status_code != 404:
        print("Failed:", filename, existing.status_code)
        continue

    result = requests.put(
        file_url,
        headers=headers,
        json=payload
    )

    if result.status_code in (200, 201):
        print("Uploaded:", filename)
    else:
        print("Failed:", filename, result.status_code,
              result.text[:200])

print("\nFinished. Check your GitHub repository.")
```
