```python
from sympy import *
z=symbols('z')
x,y=symbols('x,y',real=True)
z=x+I*y
f=z.conjugate()
u=re(f)
v=im(f)
u_dx=diff(u,x)
u_dy=diff(u,y)
v_dx=diff(v,x)
v_dy=diff(v,y)
print("u_dx : ",u_dx)
print("v_dx : ",v_dx)
print("u_dy : ",u_dy)
print("v_dy : ",v_dy)

if(u_dx==v_dy and u_dy==-v_dx):
  print("anylatic")
else:
  print("not anylatic")
```

```python
#a
from sympy import *
x, y = symbols('x y', real=True)
z = x + I*y
u = x**3 - 3*x*y**2
ux = diff(u, x)
uy = diff(u, y)
uxx = diff(ux, x)
uyy = diff(uy, y)
harmonic = simplify(uxx + uyy)
if harmonic == 0:
  print("u is harmonic")
  vx = -uy
  vy = ux
  v1 = integrate(vx, x)
  v2 = integrate(vy, y)
  v = ((v1 + v2)/2)
  print("Harmonic conjugate v(x, y):")
  print(v)
else:
  print("u is not harmonic")
```

```python
from sympy import *
t = symbols('t', real=True)
z = exp(I * t)
f = 1/z
dz_dt = diff(z, t)
integrand = f * dz_dt
integral = integrate(integrand, (t, 0, 2 * pi))
result = simplify(integral)
print(f"Contour integral of f(z) = z^2 over unit circle: {result}")
```
