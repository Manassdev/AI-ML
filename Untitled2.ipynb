```python
from sympy import Matrix
import numpy as np
A = Matrix ([[1 , 2, 3], [4, 5, 6]])
display(A)
RREF=A. rref ()
display(RREF[0])
display(RREF[1])

row_basis = RREF[0][:A. rank () , :]
display(row_basis)

col_basis = A[:, RREF[1] ]
display(col_basis)
```

```python
import sympy as sp
import numpy as np

v1 = [1, 1, 0]
v2 = [0, 1, 1]
v3 = [1, 0, 1]

V = sp.Matrix([v1, v2, v3]).T
display(V)

RREF = V.rref()
print("RREF:\n", RREF[0])
print("Pivot columns:", RREF[1])
rankV = V.rank()

if rankV == V.shape[1]:
    print("Given set is a Basis of R^3")
else:
    print("Given set is NOT a basis of R^3")
print("Dimension =", rankV)

V_np = np.array([v1, v2, v3]).T
print("NumPy Rank =", np.linalg.matrix_rank(V_np))
```

```python
import sympy as sp
import numpy as np

A = sp.Matrix([[2, 4],
               [3, 6]])
print("Matrix A:\n", A)

RREF_A = A.rref()
print("RREF:\n", RREF_A[0])
row_basis = A.rowspace()
print("Row space basis:", row_basis)
print()
```

```python
import numpy as np
A2 = np.array([[1, 2, 3, 4],
               [5, 6, 7, 8]])
rank_A2 = np.linalg.matrix_rank(A2)
print("Column space dimension =", rank_A2)
print()
```

```python
import sympy as sp
u = sp.Matrix([3, -1, 2])
v = sp.Matrix([4, 0, -3])
inner_product = u.dot(v)
print("Inner product =", inner_product)
```

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
v1 = np.array([1,0])
v2 = np.array([0,1])
plt.quiver(0,0,v1[0],v1[1],angles='xy',scale_units='xy',scale=1,color='r',label="v1")
plt.quiver(0,0,v2[0],v2[1],angles='xy',scale_units='xy',scale=1,color='b',label="v2")
plt.xlim(-2,2)
plt.ylim(-2,2)
plt.axhline(0,color='black',linewidth=0.5)
plt.axhline(0,color='black',linewidth=0.5)
plt.grid(True)
plt.gca().set_aspect('equal',adjustable='box')
plt.legend()
plt.title('span of v1 and v2')
plt.show()
```
