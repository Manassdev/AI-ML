```python

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': [],
    'F': []
}


def simple_dfs(graph, node, visited=None):

    if visited is None:
        visited = []


    if node not in visited:
        visited.append(node)


        for neighbor in graph[node]:
            simple_dfs(graph, neighbor, visited)

    return visited

result = simple_dfs(graph, 'A')
print("DFS Traversal Order:", result)
```
