```python
  from sympy import symbols ,diff
  x= symbols('x')
  f= x**4 + 3* x**2-2*x + 1
  f_prime=diff(f,x)
  print(f_prime)
```

```python
  from sympy import symbols ,diff,sin,cos
  x , y = symbols('x,y')
  f= x**3*y**2 + sin(x)*cos(y)
  df_dx=diff(f,x)
  df_dy=diff(f,y)
  print(df_dx)
  print(df_dy)
```

```python
  from sympy import symbols ,integrate,exp
  x= symbols('x')
  f= exp(2*x)+x**3
  f_i=integrate(f,x)
  print("Indefinite integral of g(x)=e^2x +x^3:")
  print(f_i)
```

```python
#24
import sympy as sp
x= sp.symbols('x')
f= x**2
def_i=sp.integrate(f,( x, 0 ,3))
print("definite integral of k(x)=x^2 from 0 to 3:")
print(def_i)
```

```python
#25
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(-5,5,100)
y=x**2-4
plt.plot(x,y,label=r'y =(x^2)-4')
plt.xlabel('x')
plt.ylabel('y')
plt.title('plot of the parabola y =(x^2)-4 ')
plt.legend()
plt.grid(True)
plt.show()
```

```python
#vector operation
#cross product
from sympy import Matrix
u =Matrix([1,2,3])
v =Matrix([4,5,6])
cross_product=u.cross(v)
print("cross_product",cross_product)
```

```python
from sympy import Matrix
A=Matrix([[1,0,1],
          [0,1,1],
          [1,1,0]])
rank_A=A.rank()
print(rank_A)
print(A.shape)
print("independent" if rank_A==A.shape[1] else "dependent")
```
